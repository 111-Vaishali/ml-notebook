# 🌸 Iris Dataset — Exploratory Data Analysis
**Dataset:** Fisher's Iris (built into scikit-learn)  
**Goal:** Understand feature distributions and class separability for classification.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

print("Shape:", df.shape)
df.head()
```

```
print("Class Distribution:")
print(df['species'].value_counts())
print("\nBasic Statistics:")
df.describe()
```

## Feature Distributions by Species

```
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
features = iris.feature_names
colors = ['#E74C3C', '#3498DB', '#2ECC71']
species = iris.target_names

for idx, (ax, feature) in enumerate(zip(axes.flatten(), features)):
    for i, sp in enumerate(species):
        subset = df[df['species'] == sp][feature]
        ax.hist(subset, bins=15, alpha=0.6, color=colors[i], label=sp, edgecolor='black')
    ax.set_title(feature, fontsize=12)
    ax.legend()

plt.suptitle('Feature Distributions by Species', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Pairplot — See All Feature Relationships at Once

```
sns.pairplot(df, hue='species', palette=['#E74C3C','#3498DB','#2ECC71'])
plt.suptitle('Iris Pairplot', y=1.02, fontsize=14)
plt.show()
```

**Key Finding:** `petal_length` and `petal_width` show very clear separation between species — they will be the most useful features for classification.


## Correlation Heatmap

```
plt.figure(figsize=(7, 5))
sns.heatmap(df.drop('species', axis=1).corr(), annot=True, fmt='.2f', cmap='YlOrRd')
plt.title('Iris Feature Correlations')
plt.tight_layout()
plt.show()
```

## Summary
- Setosa is linearly separable from the other two species
- Versicolor and Virginica overlap slightly — a good challenge for classifiers
- Petal features are more informative than sepal features
